8.b Replacing imposible values

```python
# 1. Initialize df and raw using our successfully loaded airline dataset
raw = df_airline_cleaned.copy()
df = raw.copy()

# 2. Define operational rules for the airline dataset
# For example, Delay_Minutes should be 0 or positive, and Customer_Rating should be between 1 and 5
rules = {
    "Delay_Minutes": (0, 1440),
    "Customer_Rating": (1.0, 5.0)
}

# 3. Find and replace impossible values with NaN (missing values)
for col, (low, high) in rules.items():
    if col in df.columns:
        # Find rows violating the bounds
        bad_rows = df[(df[col] < low) | (df[col] > high)].index
        if len(bad_rows) > 0:
            print(f"Replacing {len(bad_rows)} impossible values in '{col}' with NaN...")
            df.loc[bad_rows, col] = np.nan

# Print a blank line to improve readability
print()

# Count all missing cells in the original dataset vs the cleaned dataset
print("Blank cells before:", raw.isnull().sum().sum())
print("Blank cells after :", df.isnull().sum().sum())
```

```python
# Select the variable
col = "Ancillary_Revenue"

# Start with the original, uncapped data
before = raw[col].copy()

# Create a fresh dataset for capping
df_capped = raw.copy()

# Calculate Q1, Q3 and IQR using original data
Q1 = before.quantile(0.25)
Q3 = before.quantile(0.75)
IQR = Q3 - Q1

# Calculate the upper and lower limits (1.5 * IQR rule)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Apply capping to the new dataset
df_capped[col] = before.clip(lower, upper)

# Compare descriptive statistics
comparison = pd.DataFrame({
    "Before": before.describe(),
    "After": df_capped[col].describe()
})
print("Descriptive Statistics Comparison:")
print(comparison.round(1))
```

Method 2 Fill the mean

```python
import pandas as pd

# Create a copy of the dataframe for mean/mode imputation
df_mean = df.copy()

# Identify numeric columns for mean imputation
num_cols = df_mean.select_dtypes(include=['number']).columns

# Display descriptive statistics BEFORE filling missing values
print("DESCRIPTIVE STATISTICS BEFORE IMPUTATION")
display(df.describe().round(2))

# Fill missing values in numeric columns using the mean
df_mean[num_cols] = df_mean[num_cols].fillna(df_mean[num_cols].mean())

# Fill missing Route values using the most common route (mode)
df_mean["Route"] = df_mean["Route"].fillna(df_mean["Route"].mode()[0])

# Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER MEAN AND MODE IMPUTATION")
display(df_mean.describe().round(2))

# Compare the number of rows before and after
print("Rows before:", len(df))
print("Rows after :", len(df_mean))

# Count missing cells remaining after imputation
print("Blanks left:", df_mean.isnull().sum().sum())
```

Method 3 Imputation through median

```python
df_median = df.copy()
df_median[num_cols] = df_median[num_cols].fillna(df_median[num_cols].median())

# Route is text, so use the most common route
df_median["Route"] = df_median["Route"].fillna(df_median["Route"].mode()[0])

print("Blanks left:", df_median.isnull().sum().sum())
```

how did sales build week by week in each state

```python
import pandas as pd
from IPython.display import display

try:
    # 1. Ensure the 'Week' column exists in df
    if 'Week' not in df.columns:
        df['Order_Date'] = pd.to_datetime(df['Order_Date'])
        df['Week'] = df['Order_Date'].dt.isocalendar().week.astype(int)

    # 2. Compute the weekly running total of sales for each state
    running = df.pivot_table(
        index="Week",
        columns="State",
        values="Order_Value",
        aggfunc="sum"
    ).fillna(0).cumsum()

    # 3. Sort states by their final cumulative sales value
    running = running[running.iloc[-1].sort_values().index]

    print("✅ Weekly cumulative sales pivot table successfully calculated!")
    display(running)

except Exception as e:
    print(f"❌ An error occurred: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

df = pd.read_excel("/content/Zepto_Orders_Data (1).xlsx", sheet_name="Orders")
print("Rows:", df.shape[0], "| Columns:", df.shape[1])

# Ensure the 'Week' column is derived on the DataFrame
df["Week"] = df["Order_Date"].dt.isocalendar().week.astype(int)

# Running total of sales, week by week, for each state
running = df.pivot_table(
    index="Week",
    columns="State",
    values="Order_Value",
    aggfunc="sum"
).fillna(0).cumsum()

# Sort states from smallest to largest
running = running[running.iloc[-1].sort_values().index]

# Create figure
fig, ax = plt.subplots(figsize=(7, 5))

def draw(i):
    ax.clear()
    ax.barh(
        running.columns,
        running.iloc[i] / 1000,
        color="#6A1B9A"
    )
    ax.set_xlim(0, running.values.max() / 1000 * 1.1)
    ax.set_title("Sales up to week " + str(running.index[i]))
    ax.set_xlabel("Sales (INR thousand)")
    ax.grid(axis="x", alpha=0.3)

# Create animation
ani = FuncAnimation(
    fig,
    draw,
    frames=len(running),
    interval=800,
    repeat=True
)

# Display animation in Google Colab
plt.close(fig)
HTML(ani.to_jshtml())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure type_order is defined
type_order = ["New User", "Regular", "Pass Member"]

# Create the violin plot with correct column capitalization "Order_Value"
sns.violinplot(data=df, x="Customer_Type", y="Order_Value", order=type_order, color="#6A1B9A")
plt.title("Order value by customer type?")
plt.ylabel("Order Value (INR)")
plt.show()
```
