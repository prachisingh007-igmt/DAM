```python
#Zomato
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Zomato_India_Orders_Dataset - Orders_Data.csv")
```

```python
#BAR CHART
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Clean Final_Amount column (remove commas and convert to float)
df['Final_Amount'] = df['Final_Amount'].astype(str).str.replace(',', '')
df['Final_Amount'] = pd.to_numeric(df['Final_Amount'], errors='coerce')

# Step 2: Find revenue for each city
city_revenue = df.groupby("City")["Final_Amount"].sum()

# Step 3: Arrange cities from highest to lowest revenue
city_revenue = city_revenue.sort_values(ascending=False)

# Step 4: Print the revenue of each city
print(city_revenue)

# Step 5: Create a bar chart
city_revenue.plot(kind="bar")

# Step 6: Add chart title and y-axis label
plt.title("Revenue by City")
plt.ylabel("Total Revenue")

# Step 7: Show the chart
plt.show()
```

```python
#PIE CHART
import matplotlib.pyplot as plt

# Step 1: Count each payment mode
payment_counts = df["Payment_Mode"].value_counts()

# Step 2: Create a pie chart
plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct="%1.0f%%")

# Step 3: Add title
plt.title("Payment Mode Share")

# Step 4: Show the chart
plt.show()
```

```python
#PIE CHART
import matplotlib.pyplot as plt

# Step 1: Count each payment mode
payment_counts = df["Cuisine"].value_counts()

# Step 2: Create a pie chart
plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct="%1.0f%%")

# Step 3: Add title
plt.title("Cuisine Share Prachi")

# Step 4: Show the chart
plt.show()
```

```python
#HISTOGRAM
# Create histogram of order amounts
plt.hist(df["Final_Amount"].dropna(), bins=10)

# Add labels and title
plt.title("Distribution of order Amount")
plt.xlabel("Final Order Amount ($)")
plt.ylabel("Number of Order")

# Show the plot
plt.show()
```

```python
#Scatter Plot
plt.scatter(df["Distance_Km"], df["Delivery_Time_Min"])

# Add title and labels
plt.title(" Delivery Distance vs Delivery Time")
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (Mintes)")

# Show the plot
plt.show()
```

Heat Map

```python
# Heatmap: City x Cuisine order counts (the dark/zero cells show gaps)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the missing city_cuisine variable using a cross-tabulation
# Fixed: changed 'city' to 'City'
city_cuisine = pd.crosstab(df["City"], df["Cuisine"])

# Create the heatmap
# Fixed: changed 'reds' to 'Reds' with a capital 'R'
sns.heatmap(city_cuisine, annot=True, fmt="d", cmap="Reds")

# Add title
# Fixed: changed 'ply.title' to 'plt.title'
plt.title("Order: City x Cuisine (Seaborn)")

# Show the chart
plt.show()
```
